# 환율 Silver 적재

workspace.bronze.exchange_rate_koreaexim 테이블에서 메타데이터 컬럼(`_`로 시작)을 제외하고 workspace.silver.exchange_rate_koreaexim 테이블로 저장합니다.

```python
# ── Silver 스키마 생성 ────────────────────────────────
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.silver")

# ── Bronze 테이블 조회 ────────────────────────────────
bronze_df = spark.read.table("workspace.bronze.exchange_rate_koreaexim")
print("Bronze 컬럼:", bronze_df.columns)

print(f"bronze_df 대상 조회 수: {bronze_df.count()}")
# ── '_'로 시작하는 컬럼 제외, _search_date는 insert_date로 변경 ──
silver_columns = [c for c in bronze_df.columns if not c.startswith("_")]
print("Silver 대상 컬럼:", silver_columns + ["insert_date"])

from pyspark.sql.functions import col
silver_df = bronze_df.select(*silver_columns, col("_search_date").alias("insert_date"))
print(f"Silver 컬럼 수: {len(silver_columns)}")
      
print(f"Silver 대상 조회 수: {silver_df.count()}")

#a(silver_df.count())
# ── Silver 테이블에 데이터 저장 ───────────────────────
silver_df.write \
    .mode("append") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.silver.exchange_rate_koreaexim")

print(f"✅ Silver 테이블 저장 완료: workspace.silver.exchange_rate_koreaexim ({silver_df.count()}건)")
```
